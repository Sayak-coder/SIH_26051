```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv('/content/trombe_wall_dataset_categorized.csv')
df.head()
```

```python
df=df.drop(['time','best_structure','material'],axis=1)
df.head()
```

```python
df.info()
```

```python
print(df['indoor_temperature_Tin_C'].unique())
```

```python
df['indoor_temperature_Tin_C']=df['indoor_temperature_Tin_C'].map({'Below 15':0,'Between 15 to 20':1,'Between 20 to 22':2,'Between 22 to 25':3,'Above 25':4})
```

```python
print(df['indoor_temperature_Tin_C'].unique())
```

```python
df.info()
```

```python
from sklearn.model_selection import GroupShuffleSplit

X = df.drop(['indoor_temperature_Tin_C'], axis=1)
y = df['indoor_temperature_Tin_C']
groups = df['sample_id']   # keep this before you drop sample_id from X, or keep a copy

gss = GroupShuffleSplit(n_splits=1, test_size=0.2, random_state=45)
train_idx, test_idx = next(gss.split(X, y, groups=groups))

X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]
```

```python
from sklearn.preprocessing import MinMaxScaler
scaler= MinMaxScaler()
X_train_scaled = pd.DataFrame(scaler.fit_transform(X_train), columns=X_train.columns, index=X_train.index)
X_test_scaled = pd.DataFrame(scaler.transform(X_test), columns=X_test.columns, index=X_test.index)
```

```python
from sklearn.metrics import mean_squared_error
from sklearn.linear_model import LinearRegression
lrmodel=LinearRegression()
lrmodel.fit(X_train_scaled,y_train)
y_pred=lrmodel.predict(X_test_scaled)
print(mean_squared_error(y_test,y_pred))
print(lrmodel.score(X_test_scaled,y_test))
```

```python
from xgboost import XGBClassifier
xgbmodel=XGBClassifier()
xgbmodel.fit(X_train_scaled,y_train)
y_pred=xgbmodel.predict(X_test_scaled)
print(mean_squared_error(y_test,y_pred))
print(xgbmodel.score(X_test_scaled,y_test))
```
