```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv('/content/ladakh_indoor_temperature_ml_dataset.csv')
df.head()
```

```python
df.shape
```

```python
df = df.drop(['location_id', 'date', 'season'], axis=1)
df.head()
```

```python
df['best_shelter_material'].unique()
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
df['best_shelter_material']=le.fit_transform(df['best_shelter_material'])
df['best_shelter_material'].unique()
```

```python
df.info()
```

```python
from sklearn.model_selection import train_test_split
X=df.drop(['indoor_temperature_C'],axis=1)
y=df['indoor_temperature_C']

X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=45)
```

```python
from sklearn.preprocessing import MinMaxScaler
scaler=MinMaxScaler()
X_train['outdoor_temperature_C']=scaler.fit_transform(X_train[['outdoor_temperature_C']])
X_test['outdoor_temperature_C']=scaler.transform(X_test[['outdoor_temperature_C']])
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
lr_model=LinearRegression()
lr_model.fit(X_train,y_train)
y_pred=lr_model.predict(X_test)
print(r2_score(y_test,y_pred))
```

```python
from xgboost import XGBRegressor
xgb_model=XGBRegressor()
xgb_model.fit(X_train,y_train)
y_pred=xgb_model.predict(X_test)
print(r2_score(y_test,y_pred))
```
